### Iniciando as bibliotecas

```python
import pandas as pd 
import requests
import json
```

### Config para ver todas as colunas que criaremos

```python
pd.set_option('display.max_columns', None)
```

### Analisando o json inicial

```python
url = "https://rickandmortyapi.com/api/character"
response = requests.get(url)
data = response.json()

#print(data)
print(type(data))
print(data.keys())
```

### Visualizando os dados de forma "bonita"

```python
import json

# O codigo 'indent=4' cria o recuo visual (as "pastinhas")
# O 'ensure_ascii=False' serve para manter os acentos e caracteres especiais
print(json.dumps(data, indent=4, ensure_ascii=False))
```

### Mostrando apenas alguns dados dos personagens

```python
# Results é o que queremos, os personagens, o que vem antes é apenas info 
df_bruto = pd.json_normalize(data['results'])

print(df_bruto.columns)

# Pegando apenas o nome e url dos personagens 
print(df_bruto[['name','origin.name', 'location.name']].head())
```

### "Limpando" o dataset, apenas colunas específicas

```python
colunas_interesse = [
    'id', 'name', 'status', 
    'species', 'type', 'gender',
    'origin.name', 'location.name',
    'image' 
]

df_final = df_bruto[colunas_interesse].copy()

# Renomeando as tabelas
df_final.columns = [
    'id', 'nome', 'status', 'especie'
    , 'tipo', 'genero', 'origem',
    'localizacao', 'foto'
]

print("Tabela final pronta!")
display(df_final.head())
```

### Transformando o dataframe em csv

```python
# index evita criar coluna com número
df_final.to_csv('personagens_rick_morty.csv', 
                index=False, sep=';', encoding='utf-8')

print("Arquivo CSV salvo com sucesso!")
```
