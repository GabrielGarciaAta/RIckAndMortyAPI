### Importando as bibliotecas necessárias

```python
import requests
import pandas as pd
import time
```

### Configurações iniciais

```python
url_atual = "https://rickandmortyapi.com/api/character"
todos_os_personagens = []
pagina = 1

print("🛰️ Iniciando extração em massa...")
```

### Loop de extração dos personagens em todas as páginas

```python
while url_atual:
    # Requisição para a página atual
    response = requests.get(url_atual)
    data = response.json()

    # Extrai os personagens da página atual e adiciona a lista geral
    # Extend() serve pra juntar uma lista na outra 
    novos_personagens = data['results']
    todos_os_personagens.extend(novos_personagens)

    print(f"✅ Página {pagina} coletada. Total acumulado: {len(todos_os_personagens)}")

    # Pegamos o link da próxima página para o loop continuar
    url_atual = data['info']['next']
    pagina += 1

    # Pequeno intervalo para não ser bloqueado pela API
    time.sleep(0.2)

    print("\n🏁 Extração concluída")
```

### Criação do dataframe

```python
# Criando o dataframe geral e normalizando
df_completo_bruto = pd.json_normalize(todos_os_personagens)

# Seleção apenas das colunas importantes (para mim)
colunas_interesse = [
    'id', 'name', 'status', 'species', 'type', 'gender',
     'origin.name', 'location.name', 'image'
]

df_final = df_completo_bruto[colunas_interesse].copy()

# Renomeando para o português
df_final.columns = [
    'id', 'nome', 'status', 'especie', 'tipo', 'genero',
    'origem', 'localizacao', 'url_foto'   
]

print(f"Tamanho Final da tabela: {df_final.shape[0]} linhas e {df_final.shape[1]} colunas.")
display(df_final.head())
```

### Transformação do DF em arquivo CSV

```python
df_final.to_csv('../data/todos_personagens_rick_morty.csv', index=False, sep=';', encoding='utf-8')
print("💾 Arquivo CSV completo gerado na pasta/data!")
```

### Gerando o arquivo parquet usando o dataframe final

```python
# Definindo o caminho do parquet
caminho_parquet = '../data/todos_personagens.parquet'

# Função para criar o Parquet
df_final.to_parquet(caminho_parquet, index=False, compression='snappy')

print("🏛️ Arquivo Parquet gerado com sucesso!")
```
