### Importando o pandas e sqlite3

```python
import pandas as pd
import sqlite3 
```

### Lendo o parquet 

```python
df_para_sql = pd.read_parquet('../data/todos_personagens.parquet')
```

### Criando a conexão

```python
conexao = sqlite3.connect('../data/rick_and_morty.db')
```

### Injetando os dados na tabela

```python
df_para_sql.to_sql('personagens', conexao, if_exists='replace', index=False)
print("🚀 Banco de Dados SQLite criado e populado com sucesso!")
```

### Teste rápido 

```python
query = "Select nome, status, especie FROM personagens WHERE status = 'Alive' LIMIT 5"
df_teste = pd.read_sql(query, conexao)
display(df_teste)
```
