### Importando as bibliotecas necessárias

```python
import requests
import pandas as pd
import time
```

### Configurações iniciais

```python
url_atual = "https://rickandmortyapi.com/api/character"
todos_os_personagens = []
pagina = 1

print("🛰️ Iniciando extração em massa...")
```

### Loop de extração dos personagens em todas as páginas

```python
while url_atual:
    # Requisição para a página atual
    response = requests.get(url_atual)
    data = response.json()

    # Extrai os personagens da página atual e adiciona a lista geral
    # Extend() serve pra juntar uma lista na outra 
    novos_personagens = data['results']
    todos_os_personagens.extend(novos_personagens)

    print(f"✅ Página {pagina} coletada. Total acumulado: {len(todos_os_personagens)}")

    # Pegamos o link da próxima página para o loop continuar
    url_atual = data['info']['next']
    pagina += 1

    # Pequeno intervalo para não ser bloqueado pela API
    time.sleep(0.2)

    print("\n🏁 Extração concluída")
```
